```python
import pandas as pd

# Combine files without metadata
df1 = pd.read_csv('scraped_lyrics_no_metadata.csv')
df2 = pd.read_csv('scraped_lyrics_2_no_metadata.csv')
combined_no_metadata = pd.concat([df1, df2], ignore_index=True)
combined_no_metadata.to_csv('scraped_lyrics_no_metadata_combined.csv', index=False)

# Combine files with metadata
df3 = pd.read_csv('scraped_lyrics.csv')
df4 = pd.read_csv('scraped_lyrics_2.csv')
combined_with_metadata = pd.concat([df3, df4], ignore_index=True)
combined_with_metadata.to_csv('scraped_lyrics_combined.csv', index=False)
```
